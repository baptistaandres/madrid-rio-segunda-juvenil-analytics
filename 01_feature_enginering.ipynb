```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("estadisticas_equipo.csv")
df
```

```python
df["golXpart"] = (df["Goles"] / df["Part"]).round(2)
df["asistXpart"] = (df["Asist"] / df["Part"]).round(2)
df["gol/asistXpart"] = ((df['Goles'] + df['Asist']) / df['Part']).round(2)
df["%Part"] = ((df["Part"] / 30) * 100).round(1)
df["%Titu"] = ((df["Titu"] / df["Part"]) * 100).round(1)
df
```

```python
df = df.rename( columns={
    "Nombre" : "nombre",
    'Partido': 'partido',
    "Titular" : "titular",
    "Suplente" : "suplente",
    "Goles" : "goles",
    "Asist" : "asist",
    "Posicion" : "posicion",
    "golXpart" : "gol_x_part",
    "asistXpart" : "asist_x_part",
    "Contribucion_gol" : "contribucion_gol",
    "%Titularidad" : "pct_titularidad",
    "%Partidos_jugados" : "pct_partidos_jugados"
    })
df
```

```python
df.to_csv('estadisticas_equipo.csv', index=False, encoding='utf-8')
```
